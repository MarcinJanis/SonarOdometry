```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

def plot_trajectory(csv_path, start_idx=None, end_idx=None, zero_start=True):
    """
    Funkcja wczytuje plik CSV z Ground Truth i rysuje trajektorię 2D.
    
    Parametry:
    - csv_path: ścieżka do pliku (np. 'aracati/pose_gt.csv')
    - start_idx: indeks początkowy (np. 100). Jeśli None, bierze od początku.
    - end_idx: indeks końcowy (np. 500). Jeśli None, bierze do końca.
    - zero_start: czy przesunąć wykres tak, aby wybrany punkt startowy był w (0,0).
    """
    
    # 1. Wczytanie danych z CSV
    try:
        df = pd.read_csv(csv_path)
    except FileNotFoundError:
        print(f"❌ Nie znaleziono pliku: {csv_path}")
        return
    
    # 2. Wycinanie zakresu (od - do)
    # df.iloc pozwala wyciąć wiersze po ich indeksie
    df_subset = df.iloc[start_idx:end_idx].copy()
    
    if df_subset.empty:
        print("❌ Wybrany zakres nie zawiera żadnych danych!")
        return

    # 3. Pobranie współrzędnych X i Y
    x = df_subset['pose.position.x'].values
    y = df_subset['pose.position.y'].values
    
    # 4. Zerowanie początku układu (Opcjonalne, ale bardzo zalecane!)
    # Jeśli wycinamy trajektorię od klatki 500, fajnie żeby jej wykres zaczynał się w 0,0
    if zero_start:
        x = x - x[0]
        y = y - y[0]

    # 5. Rysowanie trajektorii
    plt.figure(figsize=(10, 8))
    
    # Rysowanie samej ścieżki
    plt.plot(x, y, label='Ground Truth', color='royalblue', linewidth=2, zorder=2)
    
    # Oznaczenie Startu (Zielona kropka) i Końca (Czerwona kropka)
    plt.scatter(x[0], y[0], color='lime', edgecolor='black', s=150, label='Start', zorder=3)
    plt.scatter(x[-1], y[-1], color='red', edgecolor='black', s=150, label='Koniec', zorder=3)

    # 6. Estetyka i ustawienia wykresu
    # Jeśli start_idx lub end_idx to None, pokazujemy "Początek" lub "Koniec" w tytule
    title_start = start_idx if start_idx is not None else 0
    title_end = end_idx if end_idx is not None else len(df)
    
    plt.title(f"Trajektoria Ground Truth\n(Klatki: {title_start} ➔ {title_end})", fontsize=14, fontweight='bold')
    plt.xlabel("Pozycja X [m]", fontsize=12)
    plt.ylabel("Pozycja Y [m]", fontsize=12)
    
    plt.grid(True, linestyle='--', alpha=0.6, zorder=1)
    plt.legend(fontsize=12)
    
    # KRYTYCZNE DLA TRAJEKTORII: Wymuszenie równych proporcji osi X i Y
    # Bez tego zakręt o 90 stopni mógłby wyglądać na wykresie jak zakręt o 45 stopni!
    plt.axis('equal') 
    
    plt.show()

# ==========================================
# PRZYKŁADY UŻYCIA:
# ==========================================
# Upewnij się, że podajesz właściwą nazwę pliku, który wygenerował bagpy!
moj_plik = 'C:/Users/janis/Projekty/Magisterka/SonarOdometry/data/test/aracati/ARACATI_2017_8bits_full/pose_gt.csv' # np. 'aracati/pose_gt.csv'

# # 1. Wyświetl całą trajektorię:
# plot_trajectory(moj_plik)

# 2. Wyświetl tylko fragment (np. od klatki 100 do 600):
plot_trajectory(moj_plik, start_idx=0, end_idx=4821)

# 3. Wyświetl pierwsze 500 klatek:
# plot_trajectory(moj_plik, end_idx=500)
```

```python
import pandas as pd

def convert_to_custom_format(input_csv_path, output_csv_path):
    """
    Konwertuje plik CSV z formatu ROS (bagpy) na niestandardowy format wejściowy.
    Brakujące dane (np. DVL, IMU) są wypełniane wartościami 0.0.
    """
    try:
        # 1. Wczytanie oryginalnego pliku
        df = pd.read_csv(input_csv_path)
    except FileNotFoundError:
        print(f"❌ Nie znaleziono pliku wejściowego: {input_csv_path}")
        return

    # 2. Słownik mapujący nazwy kolumn z wejścia na Twój format
    # Klucz = stara nazwa, Wartość = nowa nazwa
    rename_mapping = {
        'header.seq': 'step_idx',
        'Time': 'timestamp',
        'pose.position.x': 'pos_x',
        'pose.position.y': 'pos_y',
        'pose.position.z': 'pos_z',
        'pose.orientation.x': 'quat_x',
        'pose.orientation.y': 'quat_y',
        'pose.orientation.z': 'quat_z',
        'pose.orientation.w': 'quat_w'
    }
    
    # 3. Zmiana nazw kolumn, które istnieją w oryginalnym pliku
    df_mapped = df.rename(columns=rename_mapping)
    
    # Jeśli wolisz, żeby step_idx zawsze zaczynał się od 0 i był ciągły 
    # (niezależnie od tego co było w header.seq), odkomentuj poniższą linię:
    # df_mapped['step_idx'] = range(len(df_mapped))

    # 4. Pełna lista kolumn, których oczekuje Twój kod (w dokładnej kolejności)
    target_columns = [
        'step_idx', 'timestamp', 'pos_x', 'pos_y', 'pos_z', 
        'quat_x', 'quat_y', 'quat_z', 'quat_w', 
        'dvl_vel_x', 'dvl_vel_y', 'dvl_vel_z', 'dvl_alt', 
        'imu_orient_x', 'imu_orient_y', 'imu_orient_z', 'imu_orient_w', 
        'imu_gyro_x', 'imu_gyro_y', 'imu_gyro_z', 
        'imu_accel_x', 'imu_accel_y', 'imu_accel_z'
    ]

    # 5. Sprawdzamy, jakich kolumn z listy docelowej brakuje i wypełniamy je 0.0
    for col in target_columns:
        if col not in df_mapped.columns:
            df_mapped[col] = 0.0

    # 6. Przycięcie DataFrame tylko do docelowych kolumn i ustawienie ich w zadanej kolejności
    df_final = df_mapped[target_columns]

    # 7. Zapis do nowego pliku CSV (bez zapisywania wewnętrznego indeksu Pandasa)
    df_final.to_csv(output_csv_path, index=False)
    print(f"✅ Konwersja zakończona! Zapisano plik: {output_csv_path}")
    print(f"   Przekonwertowano {len(df_final)} wierszy.")

# ==========================================
# WYWOŁANIE FUNKCJI
# ==========================================
plik_wejsciowy = 'C:/Users/janis/Projekty/Magisterka/SonarOdometry/data/test/aracati/ARACATI_2017_8bits_full/pose_gt.csv'           # Zmień na ścieżkę do pliku z bagpy
plik_wyjsciowy = 'C:/Users/janis/Projekty/Magisterka/SonarOdometry/data/test/aracati/ARACATI_2017_8bits_full/sequence.csv'

convert_to_custom_format(plik_wejsciowy, plik_wyjsciowy)
```
