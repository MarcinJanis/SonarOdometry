```python
def visu(i_polars, row_titles=None, r_min=0.2, r_max=100.0, fov=135 * np.pi / 180, cmap = 'inferno'):
    if not isinstance(i_polars, list):
        i_polars = [i_polars]

    n = len(i_polars)

    if row_titles is not None:
        if not isinstance(row_titles, list):
            row_titles = [row_titles]
        if len(row_titles) < n:
            row_titles.extend([""] * (n - len(row_titles)))
    
    first_cart = img_polar2cart(i_polars[0], r_min, r_max, fov, out_shape=None, bg=0)
    W_pol = i_polars[0].shape[1]
    W_cart = first_cart.shape[1]

    fig, ax = plt.subplots(
        n, 2, 
        figsize=(15, 5 * n),
        squeeze=False,
        gridspec_kw={
            'width_ratios': [W_pol, W_cart],
            'wspace': 0.3,
            'hspace': 0.3
        }
    )

    for idx, i_polar in enumerate(i_polars):
        i_cart = img_polar2cart(i_polar, r_min, r_max, fov, out_shape=None, bg=0)

        H_pol, W_pol = i_polar.shape[:2]
        H_cart, W_cart = i_cart.shape[:2]

        def pol_y_pix2m(p): return r_max - (p / H_pol) * (r_max - r_min)
        def pol_y_m2pix(m): return (r_max - m) / (r_max - r_min) * H_pol

        def pol_x_pix2rad(p): return -fov/2 + (p / W_pol) * fov
        def pol_x_rad2pix(r): return (r + fov/2) / fov * W_pol

        cart_max_y = r_max * np.sin(fov/2)

        def cart_y_pix2m(p): return r_max - (p / H_cart) * r_max
        def cart_y_m2pix(m): return (r_max - m) / r_max * H_cart

        def cart_x_pix2m(p): return -cart_max_y + (p / W_cart) * (2 * cart_max_y)
        def cart_x_m2pix(m): return (m + cart_max_y) / (2 * cart_max_y) * W_cart

        # Tytuł dla całego wiersza
        if row_titles and row_titles[idx]:
            ax[idx, 0].text(-0.25, 0.5, row_titles[idx], transform=ax[idx, 0].transAxes,
                            rotation=90, va='center', ha='right', fontsize=12, fontweight='bold')

        # === Układ polarny ===
        ax[idx, 0].imshow(i_polar[:, :, 0], cmap=cmap)
        if idx == 0:
            ax[idx, 0].set_title('Układ polarny', pad=30)
        ax[idx, 0].set_xlabel('px')
        ax[idx, 0].set_ylabel('px')

        sec_x_pol = ax[idx, 0].secondary_xaxis('top', functions=(pol_x_pix2rad, pol_x_rad2pix))
        sec_x_pol.set_xlabel('theta [rad]')

        sec_y_pol = ax[idx, 0].secondary_yaxis('right', functions=(pol_y_pix2m, pol_y_m2pix))
        sec_y_pol.set_ylabel('r [m]')

        # === Układ kartezjański ===
        ax[idx, 1].imshow(i_cart[:, :, 0], cmap=cmap)
        if idx == 0:
            ax[idx, 1].set_title('Układ kartezjański', pad=30)
        ax[idx, 1].set_xlabel('px')
        ax[idx, 1].set_ylabel('px')

        sec_x_cart = ax[idx, 1].secondary_xaxis('top', functions=(cart_x_pix2m, cart_x_m2pix))
        sec_x_cart.set_xlabel('y [m]')

        sec_y_cart = ax[idx, 1].secondary_yaxis('right', functions=(cart_y_pix2m, cart_y_m2pix))
        sec_y_cart.set_ylabel('x [m]')

    plt.tight_layout(w_pad=0.05)
    plt.show()
```

## Research
According to article **["Speckle Noise Reduction in Sonar Image Based on Adaptive Redundant Dictionary"](https://www.mdpi.com/2077-1312/8/10/761#B29-jmse-08-00761)**, noise in **FLS** Sonar can be modeled as:

$$
y = x * n
$$

where $x$ is orignal Amplitude, $n$ is random noise, generated based on  **Gamma distribution**.



```python
# Read aracati img example
i_real_pth = 'C:/Users/janis/Projekty/Magisterka/SonarOdometry/data/aracati2017/fls/4000.png'
i_real = cv2.imread(i_real_pth)
i_real_rgb = cv2.cvtColor(i_real, cv2.COLOR_BGR2RGB)

h, w, _ = i_real_rgb.shape
print(f'Shape: h = {h}, w = {w}')

# Define patches to analyse
corner1 = (h // 5, w // 2)
corner2 = (3 * h // 5, w // 2 + 50)
box_h = 120
box_w = 120

color1 = (255, 50, 50)   # patch 1 color
color2 = (50, 205, 50)   # patch 2 color 

# cut out patches
patch1 = i_real_rgb[corner1[0]:corner1[0]+box_h, corner1[1]:corner1[1]+box_w, 0]
patch2 = i_real_rgb[corner2[0]:corner2[0]+box_h, corner2[1]:corner2[1]+box_w, 0]

# mark patches in fls img
I_real_marked = i_real_rgb.copy()
cv2.rectangle(I_real_marked, (corner1[1], corner1[0]), (corner1[1]+box_w, corner1[0]+box_h), color1, 3)
cv2.rectangle(I_real_marked, (corner2[1], corner2[0]), (corner2[2]+box_w if len(corner2)>2 else corner2[1]+box_w, corner2[0]+box_h), color2, 3)

# visualisation
fig = plt.figure(figsize=(14, 7))
gs = gridspec.GridSpec(2, 2, width_ratios=[3, 1], figure=fig)

ax_main = fig.add_subplot(gs[:, 0])
ax_main.imshow(I_real_marked)
# ax_main.set_title('Główny obraz z zaznaczonymi obszarami', fontsize=12, pad=10)
ax_main.axis('off')  

ax_p1 = fig.add_subplot(gs[0, 1])
ax_p1.imshow(patch1, cmap = 'gray')
ax_p1.set_title('Wycinek 1', fontsize=11, color='red', weight='bold')
for spine in ax_p1.spines.values():
    spine.set_edgecolor('red')
    spine.set_linewidth(2)

ax_p2 = fig.add_subplot(gs[1, 1])
ax_p2.imshow(patch2, cmap = 'gray')
ax_p2.set_title('Wycinek 2', fontsize=11, color='green', weight='bold')
for spine in ax_p2.spines.values():
    spine.set_edgecolor('green')
    spine.set_linewidth(2)

plt.tight_layout()
plt.show()
```

## Calculate distribution of selected region

- Calculate distribution for light and dark region.
- Fit theoretical gamma distribution parametrs for both regions.

```python
import scipy

h, w = patch1.shape

x_axis = np.linspace(0, 255, 256)

patch1_flat = patch1.flatten()
data1 =  patch1_flat 

patch2_flat = patch2.flatten()
data2 =  patch2_flat 

# ====== Distribution of patches ====== 
histogram1 = scipy.stats.rv_histogram(np.histogram(data1, bins=256, range=(0, 255), density=True))
histogram2 = scipy.stats.rv_histogram(np.histogram(data2, bins=256, range=(0, 255), density=True))

distribution1 = histogram1.pdf(x_axis)
distribution2 = histogram2.pdf(x_axis)

mean1 = histogram1.mean()
std1 = histogram1.std()

mean2 = histogram2.mean()
std2 = histogram2.std()

annotation1 = f'Wycinek 1\nμ = {mean1:.3f}, σ = {std1:.3f}'
annotation2 = f'\nWycinek 2\nμ = {mean2:.3f}, σ = {std2:.3f}'

# ==== Gamma distribution ====

shape1_fit, loc1_fit, scale1_fit = scipy.stats.gamma.fit(data1)
gamma_dist1 = scipy.stats.gamma(a=shape1_fit, loc=loc1_fit, scale=scale1_fit)

# gamma_dist1 = scipy.stats.norm(loc=loc1, scale=scale1)
shape2_fit, loc2_fit, scale2_fit = scipy.stats.gamma.fit(data2)
gamma_dist2 = scipy.stats.gamma(a=shape2_fit, loc=loc2_fit, scale=scale2_fit)


gamma1 = gamma_dist1.pdf(x_axis)
gamma2 = gamma_dist2.pdf(x_axis)

gamma1_annotation = f'\nRozkład gmmma 1\nα = {shape1_fit:.3f}, d = {loc1_fit:.3f}, β = {scale1_fit:.3}'
gamma2_annotation = f'\nRozkład gamma 2\nα = {shape2_fit:.3f}, d = {loc2_fit:.3f}, β = {scale2_fit:.3}'

# α - parametr kształtu
# d - przeusnięcie 
# β - parametr skali 


fig, ax = plt.subplots()

ax.plot(x_axis, distribution1, 'r-', label = annotation1)
ax.plot(x_axis, distribution2, 'g-', label = annotation2)

ax.plot(x_axis, gamma1, 'b--', label = gamma1_annotation)
ax.plot(x_axis, gamma2, 'k--', label = gamma2_annotation)

ax.grid(which='major', linestyle='-', linewidth='0.5', color='gray', alpha=1)
ax.grid(which='minor', linestyle='-', linewidth='0.5', color='lightgray', alpha=1)
ax.minorticks_on()

ax.set_xlabel('Wartość pikseli')
ax.set_ylabel('Gęstość prawdopodieństwa pikseli')

ax.legend()

plt.show()
```

## Common noise distribution 

As we can see, **gamma distribution** perfectly fits to distribution of sonar noise.

To obtain common distribution, independent from pixels value, we have to normalize regions.

Normalization for multiplicative nose: 

$$
n = \frac{I}{\mu}
$$

```python
# ===== Data normalization ==== 
data1_norm =  patch1_flat / np.mean(patch1_flat)
data2_norm =  patch2_flat / np.mean(patch2_flat)
data_accum = np.concat((data1_norm, data2_norm), axis=0)
# ====== Distribution of patches normnalized ====== 
max_val = max(np.max(data1_norm), np.max(data2_norm))

bins = 25

histogram_norm_1 = scipy.stats.rv_histogram(np.histogram(data1_norm, bins=bins, range=(0, max_val), density=True))
histogram_norm_2 = scipy.stats.rv_histogram(np.histogram(data2_norm, bins=bins, range=(0, max_val), density=True))
histogram_accum = scipy.stats.rv_histogram(np.histogram(data_accum, bins=bins, range=(0, max_val), density=True))


x_axis_norm = np.linspace(0, max_val, bins)

distribution_norm_1 = histogram_norm_1.pdf(x_axis_norm)
distribution_norm_2 = histogram_norm_2.pdf(x_axis_norm)
distribution_accum = histogram_accum.pdf(x_axis_norm)

mean_norm_1 = histogram_norm_1.mean()
std_norm_1 = histogram_norm_1.std()

mean_norm_2 = histogram_norm_2.mean()
std_norm_2 = histogram_norm_2.std()

mean_accum = histogram_accum.mean()
std_accum = histogram_accum.std()

annotation1 = f'Wycinek 1\nμ = {mean_norm_1:.3f}, σ = {std_norm_1:.3f}'
annotation2 = f'Wycinek 2\nμ = {mean_norm_2:.3f}, σ = {std_norm_2:.3f}'
annotation3 = f'Oba regiony\nμ = {mean_accum:.3f}, σ = {std_accum:.3f}'

# # ==== Gamma distribution ====

shape_fit, loc_fit, scale_fit = scipy.stats.gamma.fit(data_accum)

gamma_dist = scipy.stats.gamma(a=shape_fit, loc=loc_fit, scale=scale_fit)

gamma = gamma_dist.pdf(x_axis_norm)

gamma_annotation = f'Rozkład gamma\nα = {shape2_fit:.3f}, d = {loc2_fit:.3f}, β = {scale2_fit:.3}'

fig, ax = plt.subplots()

ax.plot(x_axis_norm, distribution_norm_1, 'r-', label = annotation1)
ax.plot(x_axis_norm, distribution_norm_2, 'g-', label = annotation2)
ax.plot(x_axis_norm, distribution_accum, 'b-', label = annotation3)
ax.plot(x_axis_norm, gamma, 'k--', label = gamma_annotation)

ax.grid(which='major', linestyle='-', linewidth='0.5', color='gray', alpha=1)
ax.grid(which='minor', linestyle='-', linewidth='0.5', color='lightgray', alpha=1)
ax.minorticks_on()

ax.set_xlabel('Wartość pikseli')
ax.set_ylabel('Gęstość prawdopodieństwa pikseli')
ax.legend()

plt.show()
```
